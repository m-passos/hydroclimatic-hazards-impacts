```python
import pandas as pd
from tqdm import tqdm_notebook as tqdm
import warnings
warnings.filterwarnings('ignore')
impact=pd.read_csv(r'C:\Users\MarlonVieiraPassos\OneDrive - SEI\Documents\KTH PhD\Thesis Work\Task 2\Codes\Impacts\Merged_impacts.csv')

municipalities=impact['Municipality'].unique()
impact['Time']=pd.to_datetime(impact['Time'], format='%Y')
```

```python
impact
```

```python
def interpolate(data, impact_vars, municipalities):
    all_data_imp_int=pd.DataFrame()
    for impact_var in impact_vars:
        print('Interpolating variable: '+impact_var)
        data_imp_int=pd.DataFrame()
        data_imp=data[data['Variable']==impact_var]
        with tqdm(total=(len(municipalities))) as pbar:
            for municipality in municipalities:
                data_imp_mun= data_imp[data_imp['Municipality']==municipality]
                data_imp_mun= data_imp_mun.set_index('Time')
                data_imp_mun_int = data_imp_mun.resample('D').mean()
                data_imp_mun_int['Value'] = data_imp_mun_int['Value'].interpolate()
                data_imp_mun_int['Municipality']=municipality
                data_imp_mun_int['Variable']=impact_var
                data_imp_mun_int=data_imp_mun_int.reset_index()
                data_imp_int= data_imp_int.append(data_imp_mun_int)
                pbar.update(1)
        all_data_imp_int= all_data_imp_int.append(data_imp_int)
    return all_data_imp_int
```

```python
impact_vars=['crop_yield_total', 'hydro_prod', 'discharge', 'groundwater']

int_df=interpolate(impact, impact_vars, municipalities)
```

```python
int_df
```

```python
int_df.to_csv('Merged_impacts_interpolated.csv', index=False)
```
